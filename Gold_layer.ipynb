```python
#Business KPIs

from pyspark.sql.functions import sum, count

silver_df = silver_df = spark.table("dev_catalog.pipeline_schema.silver_table")

gold_df = silver_df.groupBy("status").agg(
    count("*").alias("total_records"),
    sum("amount").alias("total_amount")
)
```

```python
gold_df.count()
gold_df.display()
```

```python
# Creating gold table

gold_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("dev_catalog.pipeline_schema.gold_table_status")
```

```python
#Daily Trends

from pyspark.sql.functions import to_date, count, sum

daily_df = silver_df.groupBy(
    to_date("updated_at").alias("date")
).agg(
    count("*").alias("records"),
    sum("amount").alias("amount")
)

display(daily_df)
```

```python
#Gold table for daily trends

daily_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("dev_catalog.pipeline_schema.gold_table_daily_trends")
```
