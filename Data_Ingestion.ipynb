```python
## Final Ingestion Code

from pyspark.sql.functions import current_timestamp, lit
import requests
url = "https://my.api.mockaroo.com/users.json?key=958a7a10"

response = requests.get(url)
data = response.json()
df = spark.createDataFrame(data)
df = df.withColumn("source", lit("api")).withColumn("ingestion_time", current_timestamp())
df.write.format("delta") \
    .mode("append") \
    .save("/Volumes/dev_catalog/pipeline_schema/bronze_table/api_data") ## writing data into broze table
```

```python
# Read back from Delta table to verify appends
table_df = spark.read.format("delta") \
    .load("/Volumes/dev_catalog/pipeline_schema/bronze_table/api_data")
    
print(f"Total rows in Delta table: {table_df.count()}")
```
